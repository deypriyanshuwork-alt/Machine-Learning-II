## 4. Logistic Regression Model

The model first calculates

$$
z = Xw+b
$$

and then

$$
\hat y = \sigma(z)
$$

where

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

If `X.shape = (m,n)` and `w.shape = (n,)`, then `X @ w` has shape `(m,)`.

## 5. Step 1 — Implement Sigmoid

### Task
Complete `sigmoid(z)`.

Use `np.exp()`. Your function should work for both a scalar and a NumPy array.

````python
def sigmoid(z):
    return 1/(1+np.exp(-z))
````

````python
test_values = np.array([-10, -2, 0, 2, 10])
print(sigmoid(test_values))

# Remember: sigmoid(0) should be 0.5
````

## 6. Step 2 — Compute the Linear Score

Implement

`z = X @ w + b`

### Hint
Use the `@` operator for matrix-vector multiplication.

````python
def compute_z(X, w, b):
    # YOUR CODE HERE
    return X@w + b
````

````python
w = np.zeros(X.shape[1])
b = 0.0

z = compute_z(X, w, b)

print("z:", z)
print("z shape:", z.shape)
````

## 7. Step 3 — Compute Predicted Probabilities

The predicted probability is

$$
\hat y = \sigma(Xw+b)
$$

### Task
Implement `predict_probability()` using the functions you have already written.

````python
def predict_probability(X, w, b):
    # YOUR CODE HERE
    temp0 = compute_z(X, w, b)
    y_hat = sigmoid(temp0)
    return y_hat
````

````python
probabilities = predict_probability(X, w, b)
print(probabilities)
````

## 8. Step 4 — Implement the Cost Function

For binary Logistic Regression,

$$
J =
-\frac{1}{m}\sum_{i=1}^{m}
\left[
y^{(i)}\log(\hat y^{(i)})
+
(1-y^{(i)})\log(1-\hat y^{(i)})
\right]
$$

Use `np.clip()` to avoid taking `log(0)`.

### Hint
```python
epsilon = 1e-15
probabilities = np.clip(probabilities, epsilon, 1-epsilon)
```

````python
def compute_cost(X, y, w, b):
    m = X.shape[0]

    # YOUR CODE HERE
    epsilon = 1e-15
    probabilities = predict_probability(X, w, b)
    probabilities = np.clip(probabilities, epsilon, 1-epsilon)
    cost = (-1/m) * (np.sum(y*np.log(probabilities) + (1-y)*np.log(1-probabilities)))

    return cost
````

````python
initial_cost = compute_cost(X, y, w, b)
print("Initial cost:", initial_cost)

# With w=0 and b=0, every probability is 0.5.
````

## 9. Step 5 — Compute the Gradients

The gradients are

$$
dw = \frac{1}{m}X^T(\hat y-y)
$$

$$
db = \frac{1}{m}\sum(\hat y-y)
$$

### Shape check

If

```text
X.shape = (m,n)
w.shape = (n,)
y.shape = (m,)
```

then

```text
X.T.shape = (n,m)
y_hat-y   = (m,)
dw        = (n,)
```

Do not use a loop over training examples. Rather, use broadcasting.

````python
def compute_gradients(X, y, w, b):
    m = X.shape[0]

    probabilities = predict_probability(X, w, b)
    dw = (1/m) * (X.T @ (probabilities - y))
    db = (1/m) * np.sum(probabilities - y)

    return dw, db
````

````python
dw, db = compute_gradients(X, y, w, b)

print("dw:", dw)
print("dw shape:", dw.shape)
print("db:", db)
````

## 10. Step 6 — Implement One Gradient-Descent Update

Use

$$
w := w - \alpha dw
$$

$$
b := b - \alpha db
$$

### Task
Compute the gradients, update `w` and `b`, and return them.

````python
def gradient_descent_step(X, y, w, b, learning_rate):
    # YOUR CODE HERE
    dw, db = compute_gradients(X, y, w, b)

    w = w - learning_rate * dw
    b = b - learning_rate * db

    return w, b
````

## 11. Step 7 — Implement Complete Gradient Descent

For each iteration:
1. Compute probabilities.
2. Compute cost.
3. Compute gradients.
4. Update parameters.
5. Store the cost.

````python
def train_logistic_regression(X, y, learning_rate=0.1, num_iterations=1000):

    w = np.zeros(X.shape[1])
    b = 0.0

    cost_history = []

    for i in range(num_iterations):

        # YOUR CODE HERE
        cost = compute_cost(X, y, w, b)
        w, b = gradient_descent_step(X, y, w, b, learning_rate)

        cost_history.append(cost)

    return w, b, cost_history
````

## 12. Train the Model

Use learning rate `0.1` and `1000` iterations.

````python
learning_rate = 0.1
num_iterations = 1000

w, b, cost_history = train_logistic_regression(
    X, y, learning_rate, num_iterations
)

print("Final weights:", w)
print("Final bias:", b)
print("Final cost:", cost_history[-1])
````

## 14. Step 8 — Convert Probabilities into Class Predictions

Use the rule

$$
\hat y =
\begin{cases}
1 & \text{if probability} \geq 0.5\\
0 & \text{otherwise}
\end{cases}
$$

### Hint

(probabilities >= threshold).astype(int)

````python
def predict(X, w, b, threshold=0.5):
    # YOUR CODE HERE
    probabilities = predict_probability(X, w, b)
    return (probabilities >= threshold).astype(int)
````

## 15. Step 9 — Calculate Accuracy

$$
Accuracy =
\frac{\text{Number of correct predictions}}
{\text{Total number of predictions}}
$$

### Task
Implement `accuracy()` using NumPy.

````python
def accuracy(y_true, y_pred):
    # YOUR CODE HERE
    return np.mean(y_true == y_pred)
````

````python
print("Accuracy:", accuracy(y, y_pred))
````

### Question
Which learning rate appears most appropriate for this dataset, and why?

````python
learning_rates = [0.001, 0.1, 1.0]

# YOUR CODE HERE
for lr in learning_rates:
    w_lr, b_lr, cost_hist_lr = train_logistic_regression(X, y, lr, 1000)
    y_pred_lr = predict(X, w_lr, b_lr)
    acc_lr = accuracy(y, y_pred_lr)
    print("lr =", lr, "-> final cost =", round(cost_hist_lr[-1], 4), ", accuracy =", acc_lr)
````

**My answer:** lr = 1.0 actually gives the best result on this dataset — lowest final cost (~0.10) and 100% accuracy — while lr = 0.001 basically doesn't move in 1000 steps (cost stays near 0.62, accuracy drops to 50%). Still, I'd call lr = 0.1 the safer general-purpose choice — this dataset is tiny (12 points) and almost linearly separable, so a large step size like 1.0 happens to work without overshooting. On bigger or noisier data that same learning rate would be a lot more likely to oscillate or diverge, so 0.1 is the more reliable pick even though 1.0 technically wins here.

Use thresholds:

Compare the predictions and accuracy.

### Question
Why can changing the threshold change the classification result without changing the learned weights?

````python
thresholds = [0.3, 0.5, 0.7]

# YOUR CODE HERE
probs = predict_probability(X, w, b)
for t in thresholds:
    y_pred_t = (probs >= t).astype(int)
    acc_t = accuracy(y, y_pred_t)
    print("threshold =", t, "-> predictions =", y_pred_t, ", accuracy =", acc_t)
````

**My answer:** w and b are already fixed after training — the threshold only decides where along the sigmoid curve we cut off and call something class 1, so changing it doesn't retrain anything. Raising the threshold to 0.7 makes the model more "confident" before it commits to class 1: the point at (3.0, 2.8) flips back to correctly being class 0, but the borderline point (3.2, 3.0) now falls just under 0.7 and gets wrongly called class 0 instead. Overall accuracy stays at 11/12 for every threshold I tried, but which single point is wrong keeps changing.
